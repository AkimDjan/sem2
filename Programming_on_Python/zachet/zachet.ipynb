```python
# Примеры кода для гифок:
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation

fig, ax = plt.subplots()
ax.set_aspect(1)

N = 64

points = ax.plot(np.linspace(-1, 1, N*4), np.linspace(-1, 1, N*4), '.')[0]

def step(t):
    x = np.sin(np.linspace(0, np.pi/2, N, endpoint=False).reshape(-1, 1) * t + np.pi/2 * np.arange(4).reshape(1, -1)).reshape(-1)
    y = np.cos(np.linspace(0, np.pi/2, N, endpoint=False).reshape(-1, 1) * t + np.pi/2 * np.arange(4).reshape(1, -1)).reshape(-1)
    points.set_data(x, y)

anim = FuncAnimation(fig, func=step, frames=np.linspace(-1, 1, 100), interval=1000/30)

plt.show()
```
