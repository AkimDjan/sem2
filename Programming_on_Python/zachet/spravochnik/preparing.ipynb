```python
def multiply_matrices_python(lhs, rhs):
    rows_lhs = len(lhs)
    cols_lhs = len(lhs[0])
    cols_rhs = len(rhs[0])
    result = [[0 for _ in range(cols_rhs)] for _ in range(rows_lhs)]
    for i in range(rows_lhs):
        for j in range(cols_rhs):
            for k in range(cols_lhs):
                result[i][j] += lhs[i][k] * rhs[k][j]
    return result

# Пример
lhs = [[1, 2,3,4,5,6,7,8], [3, 4,3,4,5,6,7,8]]
rhs = [[5, 6], [7, 8],[5, 6], [7, 8],[5, 6], [7, 8],[5, 6], [7, 8]]
result = multiply_matrices_python(lhs, rhs)
print(result)  # [[19, 22], [43, 50]]
```

```python

np.random.seed(42)

genders = np.random.choice(["men", "women"], size=200, p=[0.7, 0.3])
labels, counts = np.unique(genders, return_counts=True)

figure, axis = plt.subplots(figsize=(9, 9))
axis.set_title("Gender of Respondents", fontsize=17, fontweight="bold", c="dimgray")
wedges, _ = axis.pie(counts, explode=[0, 0.05], colors=["sandybrown", "cornflowerblue"], shadow=True)
axis.legend(wedges, labels, title="Genders")

plt.show()
```

```python
def convert_percent_to_label_str(percent: float, total: int) -> str:
    amount = int(np.round(percent / 100 * total))
    return f"{amount:d} people\n({percent:.1f}%)"

figure, axis = plt.subplots(figsize=(9, 9))
axis.set_title("Gender of Respondents", fontsize=17, fontweight="bold", c="k")
wedges, _, _ = axis.pie(
    counts,
    explode=[0, 0.05],
    autopct=lambda pct: convert_percent_to_label_str(pct, genders.size),
    colors=["sandybrown", "cornflowerblue"],
    shadow=True,
    textprops=dict(size=14, weight="bold", color="0.25")
)
axis.legend(wedges, labels, title="Genders")
plt.show()
```

```python
from matplotlib.animation import FuncAnimation
from functools import partial

abscissa = np.linspace(0, 4 * np.pi, 1000)

# Функция обновления
def update_frame(frame_id: int, *, line: plt.Line2D, abscissa: np.ndarray) -> tuple[plt.Line2D]:
    ordinates = np.sin(abscissa + frame_id * 0.1)
    line.set_ydata(ordinates)
    return line,

# Создание анимации
figure, axis = plt.subplots(figsize=(16, 9))
axis.set_xlim(abscissa.min(), abscissa.max())
axis.set_ylim(-1.5, 1.5)
axis.set_title("Sine Wave Animation", fontsize=17, fontweight="bold")
axis.set_xlabel("x", fontsize=13)
axis.set_ylabel("sin(x)", fontsize=13)
line, *_ = axis.plot(abscissa, np.sin(abscissa), c="royalblue")

animation = FuncAnimation(
    figure,
    partial(update_frame, line=line, abscissa=abscissa),
    frames=100,
    interval=10,
    blit=True
)

# Сохранение
animation.save("sine_wave.gif", writer="pillow", fps=25)
```

```python
left = pd.DataFrame({
    'CustomerID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie']
})
right = pd.DataFrame({
    'CustomerID': [2, 3, 4],
    'Country': ['UK', 'Germany', 'France']
})

# Inner merge
inner = pd.merge(left, right, on='CustomerID', how='inner')
print("Inner merge:\n", inner)

# Left merge
left_join = pd.merge(left, right, on='CustomerID', how='left')
print("\nLeft merge:\n", left_join)

# Right merge
right_join = pd.merge(left, right, on='CustomerID', how='right')
print("\nRight merge:\n", right_join)

# Outer merge
outer_join = pd.merge(left, right, on='CustomerID', how='outer')
print("\nOuter merge:\n", outer_join)
```

inner: только строки с совпадающими ключами в обеих таблицах.
left: все строки из левой таблицы, совпадающие из правой (несовпадающие заполняются NaN).
right: все строки из правой таблицы, совпадающие из левой.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# Параметры анимации
NUM_DIRECTIONS = 4       # 4 направления (верх, низ, право, лево)
POINTS_PER_ARM = 8       # Точек на каждом "луче"
FRAMES = 100             # Количество кадров
SPEED = 0.8              # Скорость движения

# Настройка фигуры
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-1.1, 1.1)
ax.set_aspect('equal')
ax.axis('off')

# Создание объекта для точек
scat = ax.scatter([], [], s=40, c='royalblue', alpha=0.8)

# Параметрические уравнения дуг
def get_trajectory(direction: str, t: float) -> tuple:
    """Возвращает координаты (x,y) для разных направлений"""
    t = np.clip(t, 0, 1)
    angle = np.pi * t
    
    if direction == 'up':
        return 0.5*np.sin(angle), 0.5*(1 - np.cos(angle))
    elif direction == 'down':
        return 0.5*np.sin(angle), -0.5*(1 - np.cos(angle))
    elif direction == 'right':
        return 0.5*(1 - np.cos(angle)), 0.5*np.sin(angle)
    elif direction == 'left':
        return -0.5*(1 - np.cos(angle)), 0.5*np.sin(angle)

# Инициализация
def init():
    scat.set_offsets(np.empty((0, 2)))
    return scat,

# Обновление кадров
def update(frame):
    positions = []
    t = frame / FRAMES * SPEED
    
    # Для каждого направления
    for direction in ['up', 'down', 'right', 'left']:
        # Базовые координаты дуги
        x_base, y_base = get_trajectory(direction, t)
        
        # Добавляем точки с небольшим случайным смещением
        for i in range(POINTS_PER_ARM):
            # Параметры для "волнового" эффекта
            phase = i * 0.08
            offset = np.random.uniform(-0.03, 0.03)
            
            x = x_base * (1 + 0.1*np.sin(t*8 + phase)) + offset
            y = y_base * (1 + 0.1*np.sin(t*8 + phase)) + offset
            
            positions.append([x, y])
    
    scat.set_offsets(positions)
    
    # Эффект исчезновения у границ
    alpha = 0.8 * (1 - t**3)
    scat.set_alpha(alpha)
    
    return scat,

# Создание анимации
ani = FuncAnimation(
    fig=fig,
    func=update,
    init_func=init,
    frames=FRAMES,
    interval=25,
    blit=True
)

# Сохранение
ani.save("square_arcs.gif", writer="pillow", fps=25, dpi=90)
plt.close()

print("Готово! Анимация сохранена как square_arcs.gif")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from functools import partial

n_bins = 50
abscissa = np.linspace(-1, 1, n_bins)
frames_per_cycle = 100
fig, ax = plt.subplots(figsize=(16, 9))
ax.set_xlim(-1.1, 1.1)
ax.set_ylim(-0.1, 2.1)
ax.set_title("50 столбцов, Анимация меняется от y=x**3 до y=-x**3", fontsize=17, fontweight="bold")
ax.set_xlabel("x", fontsize=13)
ax.set_ylabel("y", fontsize=13)

dx = abscissa[1] - abscissa[0]
width = dx * 0.8
bars = ax.bar(abscissa, 1+abscissa**3, width=width, align='center')

def update_frame(frame, bars, abscissa, frames_per_cycle):
    t = 0.5 * (1 - np.cos(2 * np.pi * frame / frames_per_cycle))
    new_heights = 1 + abscissa**3 * (1 - 2 * t)
    for i, bar in enumerate(bars):
        bar.set_height(new_heights[i])
    return tuple(bars)

animation = FuncAnimation(
    fig,
    partial(update_frame, bars=bars, abscissa=abscissa, frames_per_cycle=frames_per_cycle),
    frames=frames_per_cycle,
    interval=20,
    blit=True
)

animation.save("cubic_histogram.gif", writer="pillow", fps=30)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from functools import partial

plt.style.use("ggplot")


def update_frame(frame, lines, theta, all_radii, rotation_speed=1):
    updated = []
    for idx, line in enumerate(lines):
        start_idx = idx * (all_radii.shape[0] // len(lines))

        current_idx = (start_idx + frame) % all_radii.shape[0]
        current_radius = all_radii[current_idx]
        # print(current_idx, current_radius)

        angle = -frame * rotation_speed

        x = current_radius * np.cos(theta + angle)
        y = current_radius * np.sin(theta + angle)

        line.set_data(x, y)
        updated.append(line)
    return tuple(updated)


fig, ax = plt.subplots(figsize=(8, 8))
ax.set_aspect('equal')
# ax.set_xlim(-15, 15)
# ax.set_ylim(-15, 15)
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axis('off')

n_radii = 100
min_radius = 1
max_radius = 12
all_radii = np.linspace(min_radius, max_radius, n_radii)

n_circles = 5
n_points = 100
theta = np.linspace(0, 2 * np.pi, n_points)
color = 'royalblue'


lines = []
for i in range(n_circles):
    initial_idx = i * (len(all_radii) // n_circles)
    initial_radius = all_radii[initial_idx]
    x = initial_radius * np.cos(theta)
    y = initial_radius * np.sin(theta)
    line, = ax.plot(x, y, 'o', markersize=4, color=color)
    lines.append(line)

animation = FuncAnimation(
    fig,
    partial(update_frame, lines=lines, theta=theta, all_radii=all_radii),
    frames=500,
    interval=50,
    blit=True,
    repeat=True
)

animation.save("z_continuous_radius_cycle.gif",
               writer="pillow",
               fps=15,
               dpi=100,
               savefig_kwargs={'facecolor': 'black'})
```
