```python
import numpy as np
from matplotlib import pyplot as plt
import math
```

```python
def Poisson(x, Lambda):
    return Lambda**x * np.exp(-Lambda) / math.factorial(x)
```

```python
def gauss(x, mu, sigma):
    return 1 / (np.sqrt(2 * np.pi) * sigma) * np.exp(-(x - mu)**2 / (2 * sigma**2))
```

```python
def integral(Y, h):
    return np.sum(Y * h)
```

```python
mu = 24.86
sigma = 4.84
X = np.linspace(mu - 2*sigma, mu + 2*sigma, 201)
Y = gauss(X, mu, sigma)
```

```python
plt.plot(X, Y)
```

```python
integral(Y, 4*sigma / 201)
```

```python
Poisson_X = np.linspace(0, 50, 51, dtype=int)
Poisson_Y = np.zeros_like(Poisson_X, dtype=float)
Poisson_Y1 = np.zeros_like(Poisson_X, dtype=float)
for i in range(Poisson_X.shape[0]):
    Poisson_Y[i] = Poisson(Poisson_X[i], mu)
    Poisson_Y1[i] = Poisson(Poisson_X[i], sigma**2)
```
